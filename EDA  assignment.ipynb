```python
plt.figure(figsize=(15,5))
x=df.groupby("Category")['Unit_Price'].mean()
plt.bar(x.index,x.values)
plt.title('CATEGORY v/s Unit_Price')
plt.xlabel('Category')
plt.ylabel('Unit_Price')
plt.show()
```

```python
x=df.groupby('Category')['Quantity'].mean()
plt.bar(x.index ,x.values)
plt.title("CATEGORY v/s QUANTITY")
plt.xlabel('CATEGORY')
plt.ylabel('QUANTITY')
plt.show()
```

```python
plt.figure(figsize=(30,10))
x=df.groupby("Product")['Discount'].mean()
plt.bar(x.index,x.values)
plt.title('Product v/s Discount')
plt.xlabel('Product')
plt.ylabel('Discount')
plt.show()
```

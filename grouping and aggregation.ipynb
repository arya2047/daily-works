```python
import pandas as pd
```

```python
data={
    'employee':['amit','arya','minnu','ammu','anu','aathi'],
    'department':['IT','HR','IT','FINANCE','HR','IT'],
    'city':['calicut','chennai','delhi','mumbai','chennai','calicut'],
    'salary':[70000,30000,100000,40000,80000,90000],
    'experience':[1,4,5,8,9,4]

}
df=pd.DataFrame(data)
df
```

```python
df.groupby('department')
```

```python
df.groupby('department')['salary'].mean()
```

```python
df.groupby('department')['salary'].sum()
```

```python
df.groupby('department')['experience'].max()

```

```python
df.groupby('department')['salary'].agg(['mean','sum','min','max'])
```

```python
df.groupby('department').agg({
    'salary':['mean','sum'],
    'experience':['mean','max']
})
```

```python
df.groupby('department').agg(
    avg_salary=('salary','mean'),
    total_salary=('salary','sum'),
    max_experience=('experience','max')
)
```

```python
pd.merge(employees,departments,on="deptID",how='right')
```

```python

data={
    'ID':[1,2,3,4,2,5],
    'NAME':['ali','john','sara','None','john','david'],
    'AGE':[24,np.nan,22,20,np.nan,34],
    'SALARY':[50000,40000,None,70000,30000,20000],
    'DEPARTMENT':['HR','IT','FINANCE',None,'IT','IT']

}

df=pd.DataFrame(data)
df
```

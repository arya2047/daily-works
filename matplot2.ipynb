```python
data=[2,4,5,6,7,5,40,35,46,27,35,87,87,57,89,500,1000,-1,]
plt.boxplot(data)
plt.title("box plot")
plt.show()
```

```python
days=[1,2,3,4,5]
sales=[100,200,400,700,100]
plt.fill_between(days,sales,color='lightgreen')
plt.title("area plot")
plt.show()
```

```python
months=['jan','feb','mar','apr','may','jun']

electronics=[120,30,58,90,40,25]
furniture=[20,35,70,15,45,100]
clothing=[100,200,300,400,700,400]

plt.stackplot(
    months,
    electronics,
    furniture,
    clothing,
    labels=['electronics','furniture','clothing']
)
plt.xlabel('month')
plt.ylabel("sales,(₹ thousands)")
plt.title('monthly sales by product category')

plt.legend()
plt.show()
```

```python
plt.plot(months,electronics)
plt.text('mar',90,'plot')
plt.show()
```
